## I. What is Pandas?

**Pandas** is the most widely used Python library for **data analysis and manipulation**.
It gives you powerful tools to load, clean, explore, and transform data — all in just a few lines of code.

Think of it as a **supercharged spreadsheet** you can program.

---

### Why use Pandas?

| Task | Without Pandas | With Pandas |
|------|---------------|-------------|
| Load a CSV file | Many lines of file parsing | `pd.read_csv('file.csv')` |
| Filter rows | Loops + conditions | `df[df['Age'] > 18]` |
| Average of a column | Manual sum/len | `df['Grade'].mean()` |
| Group and summarise | Nested dicts + loops | `df.groupby('Course').mean()` |
| Handle missing values | Lots of `if None` checks | `df.fillna(0)` |

---

### Two key data structures

Pandas is built around two core structures:

| Structure | Dimensions | Think of it as... |
|-----------|------------|-------------------|
| **Series** | 1D (one column) | A single column in a spreadsheet |
| **DataFrame** | 2D (rows & columns) | A full spreadsheet / table |

## III. Series

### What is a Series?

A **Series** is a **one-dimensional** labelled array. It is like a single column from a spreadsheet.

```
Index   Value
  0  →   10
  1  →   20
  2  →   30
  3  →   40
```

Every element has:
- A **value** — the actual data
- An **index** (label) — used to access that value

By default, the index is `0, 1, 2, ...` — but you can set **custom labels**.

### Creating a Series

#### a. From a list (default numeric index)

````python
import pandas as pd

grades = pd.Series([78, 85, 91, 62, 95])
print(grades)
````

#### b. From a list with a custom index

````python
grades = pd.Series(
    [78, 85, 91, 62, 95],
    index=['Alice', 'Bob', 'Diana', 'Charlie', 'Iris']
)
print(grades)
````

#### c. From a dictionary (keys become the index automatically)

````python
scores = pd.Series({'Maths': 88, 'Physics': 76, 'CS': 92})
print(scores)
````

### Accessing Elements in a Series

| Method | What it does |
|--------|--------------|
| `s['label']` | Access by index label |
| `s[0]` | Access by position (like a list) |
| `s.iloc[0]` | Access by position (explicit) |
| `s[1:3]` | Slice by position |

````python
grades = pd.Series(
    [78, 85, 91, 62, 95],
    index=['Alice', 'Bob', 'Diana', 'Charlie', 'Iris']
)

print('Alice grade:', grades['Alice'])     # by label
print('First grade:', grades.iloc[0])      # by position
print('\nFirst three students:\n', grades[:3])  # slice
````

### Series Operations

You can do maths on a whole Series at once — no loops needed!

````python
grades = pd.Series([78, 85, 91, 62, 95], index=['Alice', 'Bob', 'Diana', 'Charlie', 'Iris'])

print('Average grade:   ', grades.mean())
print('Highest grade:   ', grades.max())
print('Lowest grade:    ', grades.min())
print('Total students:  ', grades.count())

# Add 5 bonus points to everyone
print('\nWith +5 bonus:\n', grades + 5)
````

### Filtering a Series

You can filter a Series using a **condition** — just like filtering a list, but much simpler.

````python
grades = pd.Series([78, 85, 91, 62, 95], index=['Alice', 'Bob', 'Diana', 'Charlie', 'Iris'])

# Students with grade above 80
high_achievers = grades[grades > 80]
print('Students with grade > 80:\n', high_achievers)
````

## IV. DataFrame

### What is a DataFrame?

A **DataFrame** is a **two-dimensional** table with rows and columns — exactly like a spreadsheet.

```
       Name   Age       City
  0   Alice    19     London       ← row 0
  1     Bob    21  Manchester      ← row 1
  2   Diana    22  Birmingham      ← row 2
  ↑               ↑
index          columns
```

- Each **column** is a `Series`.
- Each **row** represents one record.
- The **index** labels the rows (default: 0, 1, 2, ...).

### Creating a DataFrame

#### a. From a dictionary (most common)

Each **key** becomes a column name, each **list** becomes the column data.

````python
import pandas as pd

data = {
    'Name':   ['Alice', 'Bob', 'Diana', 'Charlie'],
    'Age':    [19, 21, 22, 20],
    'Course': ['Computer Science', 'Data Science', 'Mathematics', 'Physics'],
    'Grade':  [78, 85, 91, 62]
}

df = pd.DataFrame(data)
df
````

#### b. From a list of dictionaries

Each dictionary becomes one row.

````python
students = [
    {'Name': 'Alice',   'Age': 19, 'Grade': 78},
    {'Name': 'Bob',     'Age': 21, 'Grade': 85},
    {'Name': 'Diana',   'Age': 22, 'Grade': 91},
]

df2 = pd.DataFrame(students)
df2
````

### DataFrame Properties

After creating or loading a DataFrame, these **attributes** give you quick information:

| Attribute / Method | What it returns |
|--------------------|-----------------|
| `df.shape` | (rows, columns) as a tuple |
| `df.ndim` | Number of dimensions (always 2) |
| `df.size` | Total number of cells (rows × columns) |
| `df.columns` | List of column names |
| `df.index` | Row labels |
| `df.dtypes` | Data type of each column |
| `df.info()` | Full summary: non-null counts, dtypes, memory |

````python
data = {
    'Name':   ['Alice', 'Bob', 'Diana', 'Charlie'],
    'Age':    [19, 21, 22, 20],
    'Course': ['Computer Science', 'Data Science', 'Mathematics', 'Physics'],
    'Grade':  [78, 85, 91, 62]
}
df = pd.DataFrame(data)

print('Shape:  ', df.shape)       # (4, 4)
print('Size:   ', df.size)        # 16
print('Columns:', list(df.columns))
print()
print('Data types:\n', df.dtypes)
````

### Selecting Rows with `iloc` (position-based)

`iloc` selects rows and columns **by their position** (like list indexing).

```python
df.iloc[row_position, column_position]
df.iloc[start:stop]          # slice rows
df.iloc[start:stop, 0:3]     # slice rows AND columns
```

````python
df = pd.read_csv('week07_data/students.csv')

print('=== First row (position 0) ===')
print(df.iloc[0])

print('\n=== Rows 2 to 4 ===')
print(df.iloc[2:5])

print('\n=== Last row ===')
print(df.iloc[-1])
````

### Selecting Rows with `loc` (label-based)

`loc` selects rows and columns **by label** (index label or column name).
It's also the standard way to **update specific values**.

```python
df.loc[row_label, 'column_name']
df.loc[condition, 'column_name']   # select rows matching a condition
```

````python
df = pd.read_csv('week07_data/students.csv')

# Select row at index 0 (default index is 0,1,2...)
print(df.loc[0])

print()

# Select specific columns for a specific row
print(df.loc[0, 'Name'])      # 'Alice Johnson'
print(df.loc[0, 'Grade'])     # 78
````

## VIII. Filtering Data

**Filtering** means selecting only the rows that meet a condition.
This is one of the most powerful features of Pandas.

### How it works — Boolean Indexing

A condition on a column produces a **True/False Series**, and Pandas keeps only the `True` rows.

```python
df['Grade'] > 80
# 0     True
# 1     True
# 2    False
# 3     True
# dtype: bool

df[df['Grade'] > 80]   # keep only True rows
```

### Single Condition

````python
df = pd.read_csv('week07_data/students.csv')

# Students with Grade above 85
high_grades = df[df['Grade'] > 85]
print(f'Students with Grade > 85: {len(high_grades)}')
high_grades
````

### Multiple Conditions

Combine conditions using:
- `&` — AND (both must be True)
- `|` — OR (either can be True)
- `~` — NOT (flip True/False)

> **Important:** Always wrap each condition in **parentheses** `( )` when combining.

````python
df = pd.read_csv('week07_data/students.csv')

# AND: Computer Science students with Grade above 75
cs_high = df[(df['Course'] == 'Computer Science') & (df['Grade'] > 75)]
print('CS students with Grade > 75:')
print(cs_high[['Name', 'Grade', 'Course']])
````

````python
df = pd.read_csv('week07_data/students.csv')

# OR: students from London OR Birmingham
lon_birm = df[(df['City'] == 'London') | (df['City'] == 'Birmingham')]
print(f'Students from London or Birmingham: {len(lon_birm)}')
print(lon_birm[['Name', 'City']].head(8))
````

### Filtering with `isin()` — cleaner OR for many values

Instead of chaining many `|` conditions, use `isin()` with a list of values.

````python
df = pd.read_csv('week07_data/students.csv')

# Students studying CS or Data Science (cleaner than using |)
stem = df[df['Course'].isin(['Computer Science', 'Data Science'])]
print(f'CS or Data Science students: {len(stem)}')
stem[['Name', 'Course', 'Grade']].head(8)
````

## IX. Modifying Data

### Adding a New Column

Assign a value or calculation to a new column name — Pandas creates it automatically.

````python
df = pd.read_csv('week07_data/students.csv')

# Add a Pass/Fail column
df['Result'] = df['Grade'].apply(lambda g: 'Pass' if g >= 60 else 'Fail')

df[['Name', 'Grade', 'Result']].head(8)
````

### Calculated Column (arithmetic)

You can create a new column from arithmetic on existing columns.

````python
df = pd.read_csv('week07_data/students.csv')

# Combined score: 70% grade + 30% attendance
df['Score'] = (df['Grade'] * 0.7 + df['Attendance'] * 0.3).round(1)

df[['Name', 'Grade', 'Attendance', 'Score']].head(6)
````

### Updating Values with `loc`

Use `loc` to change specific values — rows matching a condition in a specific column.

````python
df = pd.read_csv('week07_data/students.csv')

# Change the City for 'Bob Smith' to 'Leeds'
df.loc[df['Name'] == 'Bob Smith', 'City'] = 'Leeds'

print(df[df['Name'] == 'Bob Smith'][['Name', 'City']])
````

### Renaming Columns

Use `df.rename()` with a dictionary mapping old name → new name.

````python
df = pd.read_csv('week07_data/students.csv')

df = df.rename(columns={'Grade': 'Mark', 'Attendance': 'Att_%'})
print(df.columns.tolist())
````

### Dropping Columns or Rows

Use `df.drop()` with `axis=1` for columns and `axis=0` for rows.

````python
df = pd.read_csv('week07_data/students.csv')

# Drop the Scholarship column
df_no_scholar = df.drop('Scholarship', axis=1)
print(df_no_scholar.columns.tolist())

# Drop row at index 0
df_no_first = df.drop(0, axis=0)
print(f'\nRows before: {len(df)}, After: {len(df_no_first)}')
````

## X. Handling Missing Data

Real datasets almost always have **missing values** — represented in Pandas as `NaN` (Not a Number).

```
   Name  Age   Grade
0  Alice   19    78.0
1    Bob   21     NaN   ← missing grade
2  Diana  NaN    91.0   ← missing age
```

Handling missing data is a **critical step** in any data analysis — if you ignore it, calculations will be wrong.

### Step 1 — Detect Missing Values

````python
df = pd.read_csv('week07_data/students_missing.csv')

print('Missing values per column:')
print(df.isnull().sum())
````

````python
df = pd.read_csv('week07_data/students_missing.csv')

# Show only rows that have at least one missing value
rows_with_missing = df[df.isnull().any(axis=1)]
rows_with_missing
````

### Step 2 — Fill Missing Values with `fillna()`

Replace `NaN` with a specific value, or with a calculated value like the column mean.

````python
df = pd.read_csv('week07_data/students_missing.csv')

# Fill missing Grade with the column average
avg_grade = df['Grade'].mean()
df['Grade'] = df['Grade'].fillna(avg_grade)

# Fill missing Scholarship with 'Unknown'
df['Scholarship'] = df['Scholarship'].fillna('Unknown')

print('Missing values after fillna:')
print(df.isnull().sum())
````

### Step 3 — Drop Rows/Columns with `dropna()`

Sometimes it's better to **remove** rows with missing data entirely.

````python
df = pd.read_csv('week07_data/students_missing.csv')

print(f'Rows before dropna: {len(df)}')

# Drop rows where Attendance is missing
df_clean = df.dropna(subset=['Attendance'])

print(f'Rows after dropna:  {len(df_clean)}')
````

## XI. Statistics & Sorting

### Aggregation Functions

These work on a **column (Series)** and return a single number:

| Method | Returns |
|--------|---------|
| `.mean()` | Average |
| `.sum()` | Total |
| `.max()` | Maximum value |
| `.min()` | Minimum value |
| `.count()` | Count of non-missing values |
| `.std()` | Standard deviation |
| `.median()` | Middle value |

````python
df = pd.read_csv('week07_data/students.csv')

print(f'Average Grade:    {df["Grade"].mean():.2f}')
print(f'Highest Grade:    {df["Grade"].max()}')
print(f'Lowest Grade:     {df["Grade"].min()}')
print(f'Median Grade:     {df["Grade"].median()}')
print(f'Std Dev:          {df["Grade"].std():.2f}')
print(f'Total students:   {df["Grade"].count()}')
````

### Sorting with `sort_values()`

Sort by one or more columns. `ascending=False` means highest first.

````python
df = pd.read_csv('week07_data/students.csv')

# Top 5 students by Grade
top5 = df.sort_values('Grade', ascending=False).head(5)
top5[['Name', 'Course', 'Grade']]
````

````python
df = pd.read_csv('week07_data/students.csv')

# Sort by Course (A-Z), then by Grade (highest first) within each course
df_sorted = df.sort_values(['Course', 'Grade'], ascending=[True, False])
df_sorted[['Name', 'Course', 'Grade']].head(10)
````

## XII. Groupby

**Groupby** splits the data into groups, applies a function, and combines the results.
Think of it as the Pandas version of Excel's **PivotTable**.

```
df.groupby('Course')['Grade'].mean()
       ↑                ↑       ↑
  split by         column    aggregate
```

### Basic Groupby

````python
df = pd.read_csv('week07_data/students.csv')

# Average grade per course
avg_by_course = df.groupby('Course')['Grade'].mean().round(2)
print('Average Grade per Course:')
print(avg_by_course.sort_values(ascending=False))
````

````python
df = pd.read_csv('week07_data/students.csv')

# Number of students per course
students_per_course = df.groupby('Course')['Name'].count()
print('Students per Course:')
print(students_per_course)
````

### Multiple Aggregations with `agg()`

Calculate several statistics at once using a dictionary.

````python
df = pd.read_csv('week07_data/students.csv')

summary = df.groupby('Course').agg(
    Count       = ('Name',       'count'),
    Avg_Grade   = ('Grade',      'mean'),
    Max_Grade   = ('Grade',      'max'),
    Avg_Attend  = ('Attendance', 'mean')
).round(2)

summary.sort_values('Avg_Grade', ascending=False)
````

## XIII. Saving Data

After cleaning or analysing data, you often want to **save** the result.

| Method | Format |
|--------|--------|
| `df.to_csv('file.csv', index=False)` | CSV |
| `df.to_excel('file.xlsx', index=False)` | Excel |
| `df.to_json('file.json')` | JSON |

> **Note:** `index=False` prevents Pandas from writing the row numbers (0, 1, 2…) as an extra column.

````python
df = pd.read_csv('week07_data/students.csv')

# Filter high achievers and save to a new file
high_achievers = df[df['Grade'] > 80]
high_achievers.to_csv('week07_data/high_achievers.csv', index=False)

print(f'Saved {len(high_achievers)} students to high_achievers.csv')

# Verify by reading it back
check = pd.read_csv('week07_data/high_achievers.csv')
check.head()
````
