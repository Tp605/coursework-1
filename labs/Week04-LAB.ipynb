### Task 6: Classify Server Log Entries

Write a function `classify_logs(logs)` that takes a list of log strings and:
1. Counts how many `ERROR`, `INFO`, and `WARNING` entries there are
2. Extracts and prints the **timestamp** from each entry
3. Returns the three counts

**Methods to use:** `startswith()`, `split()`, `strip()`


```python
logs = [
    "[14:32:05] ERROR: Disk full",
    "[14:32:10] INFO: Server started",
    "[14:33:01] WARNING: High memory usage",
    "[14:33:15] INFO: Backup complete",
    "[14:34:00] ERROR: Connection refused",
    "[14:35:22] WARNING: CPU at 90%"
]

def classify_logs(logs):

    error_count = 0
    info_count = 0
    warning_count = 0

    for log in logs:
        log = log.strip()
        # Extract timestamp
        timestamp = log.split("]")[0] + "]"
        print("Timestamp:", timestamp)

        # Count log types
        if "ERROR" in log:
            error_count += 1
        elif "INFO" in log:
            info_count += 1
        elif "WARNING" in log:
            warning_count += 1

    return error_count, info_count, warning_count

errors, infos, warnings = classify_logs(logs)

print("\nERROR logs:", errors)
print("INFO logs:", infos)
print("WARNING logs:", warnings)
```

### Task 7: Simple Sentiment Analyser

Write a function `analyse_sentiment(reviews)` that:
1. Cleans each review (`strip()`, `lower()`)
2. Classifies it as **positive**, **negative**, or **neutral** using keyword lists
3. Returns a dictionary with counts and a list of which reviews are negative

**Methods/Functions to use:** `strip()`, `lower()`, `any()`, `in`


```python
reviews = [
    "  AMAZING product, Love it!!!  ",
    "terrible experience   ",
    "  It was okay  ",
    "WORST purchase ever",
    "  fantastic quality  ",
    "poor customer service",
    "  Not bad, not great  "
]

def analyse_sentiment(reviews):

    positive_words = ["amazing", "love", "fantastic", "great"]
    negative_words = ["terrible", "worst", "poor", "bad"]

    results = {
        "positive": 0,
        "negative": 0,
        "neutral": 0,
        "negative_reviews": []
    }

    for review in reviews:

        # Clean review
        cleaned = review.strip().lower()

        # Check positive
        if any(word in cleaned for word in positive_words):

            results["positive"] += 1

        # Check negative
        elif any(word in cleaned for word in negative_words):

            results["negative"] += 1

            results["negative_reviews"].append(cleaned)

        # Otherwise neutral
        else:
            results["neutral"] += 1

    return results


# Function call
output = analyse_sentiment(reviews)

print(output)
```

### Task 8: Data Validation Pipeline

Write a function `validate_record(record)` that takes a CSV string and checks:
1. It has exactly 4 fields (name, age, email, phone)
2. Name contains only letters and spaces
3. Age is a valid number between 18 and 100
4. Email contains `@` and `.`
5. Phone starts with `+` or `0`

Return a list of **error messages** (empty list = valid record).

Use `all()` to check if ALL fields are non-empty first.

**Methods to use:** `split()`, `isdigit()`, `isalpha()`, `startswith()`, `all()`


```python
records = [
    "Alice Smith,25,alice@email.com,+447123456789",
    "Bob,abc,bob@email,07123456789",
    "Charlie123,30,charlie@mail.com,447123456789",
    ",22,missing@mail.com,+440000000000"
]

def validate_record(record):

    errors = []

    # Split CSV fields
    fields = record.split(",")

    # Check number of fields
    if len(fields) != 4:
        errors.append("Record must contain 4 fields")
        return errors

    # Remove spaces
    name = fields[0].strip()
    age = fields[1].strip()
    email = fields[2].strip()
    phone = fields[3].strip()

    # Check all fields are not empty
    if not all([name, age, email, phone]):
        errors.append("All fields must be filled")

    # Validate name
    if not all(part.isalpha() for part in name.split()):
        errors.append("Name must contain only letters and spaces")

    # Validate age
    if not age.isdigit():
        errors.append("Age must be a valid number")

    else:
        age_number = int(age)

        if age_number < 18 or age_number > 100:
            errors.append("Age must be between 18 and 100")

    # Validate email
    if "@" not in email or "." not in email:
        errors.append("Invalid email address")

    # Validate phone
    if not (phone.startswith("+") or phone.startswith("0")):
        errors.append("Phone must start with + or 0")

    return errors


# Test records
for record in records:

    result = validate_record(record)

    print("\nRecord:", record)

    if result == []:
        print("Valid Record")

    else:
        print("Errors:")

        for error in result:
            print("-", error)
```

### Task 9: Word Frequency Counter

Write a function `word_frequency(text)` that:
1. Converts text to lowercase
2. Removes punctuation (`.`, `,`, `!`, `?`)
3. Splits into words
4. Returns a **dictionary** of word → count
5. Print the top 5 most common words

**Methods to use:** `lower()`, `replace()`, `split()`, `count()`


```python
text = """Data science is the study of data. Data scientists use data to find patterns.
Machine learning is a part of data science. Scientists study data every day!
The science of data is growing, and data is everywhere."""

def word_frequency(text):

    # Convert to lowercase
    text = text.lower()

    # Remove punctuation
    text = text.replace(".", "")
    text = text.replace(",", "")
    text = text.replace("!", "")
    text = text.replace("?", "")

    # Split into words
    words = text.split()

    # Dictionary to store counts
    frequency = {}

    # Count words
    for word in words:

        frequency[word] = words.count(word)

    return frequency


# Function call
result = word_frequency(text)

print("Word Frequencies:\n")

for word, count in result.items():
    print(word, ":", count)


# Top 5 common words
print("\nTop 5 Most Common Words:\n")

sorted_words = sorted(result.items(), key=lambda item: item[1], reverse=True)

for word, count in sorted_words[:5]:
    print(word, ":", count)
```

### Task 10: Password Strength Checker

Write a function `check_password(password)` that scores a password (0–5):

| Rule | Points |
|------|--------|
| At least 8 characters | +1 |
| Contains uppercase letter | +1 |
| Contains lowercase letter | +1 |
| Contains a digit | +1 |
| Contains a special character (`@#$%!`) | +1 |

Print the score and a label: **Weak** (0-2), **Medium** (3-4), **Strong** (5).

If the password contains a space, print  and reject it.

**Methods to use:** `len()`, `isupper()`, `islower()`, `isdigit()`, `any()`


```python
test_passwords = [
    "hello",
    "Hello123",
    "My Pass!",
    "Str0ng@Pass",
    "12345678"
]

def check_password(password):

    score = 0

    # Reject spaces
    if " " in password:
        print(password, "-> Invalid password (contains spaces)")
        return

    # Rule 1: At least 8 characters
    if len(password) >= 8:
        score += 1

    # Rule 2: Uppercase letter
    if any(char.isupper() for char in password):
        score += 1

    # Rule 3: Lowercase letter
    if any(char.islower() for char in password):
        score += 1

    # Rule 4: Digit
    if any(char.isdigit() for char in password):
        score += 1

    # Rule 5: Special character
    if any(char in "@#$%!" for char in password):
        score += 1

    # Strength label
    if score <= 2:
        strength = "Weak"

    elif score <= 4:
        strength = "Medium"

    else:
        strength = "Strong"

    print(password, "-> Score:", score, "|", strength)


# Test passwords
for pwd in test_passwords:
    check_password(pwd)
```

### Task 11: Caesar Cipher Encoder & Decoder

Write **two** functions:
1. `encode(text, key)` — shifts each letter forward by `key` positions
2. `decode(text, key)` — reverses the shift to get the original text

Rules:
- Only shift letters (a-z). Leave spaces, digits, and punctuation unchanged
- Wrap around: `z` shifted by 1 → `a`
- Preserve case: uppercase stays uppercase
- Verify: `decode(encode(text, key), key)` should return the original text

**Methods to use:** `ord()`, `chr()`, `isalpha()`, `isupper()`, `islower()`


```python
# Function to encode text
def encode(text, key):

    result = ""

    for char in text:

        # Check letters only
        if char.isalpha():

            # Uppercase letters
            if char.isupper():

                new_char = chr((ord(char) - ord('A') + key) % 26 + ord('A'))

            # Lowercase letters
            else:

                new_char = chr((ord(char) - ord('a') + key) % 26 + ord('a'))

            result += new_char

        # Keep spaces, digits, punctuation same
        else:
            result += char

    return result


# Function to decode text
def decode(text, key):

    result = ""

    for char in text:

        if char.isalpha():

            if char.isupper():

                new_char = chr((ord(char) - ord('A') - key) % 26 + ord('A'))

            else:

                new_char = chr((ord(char) - ord('a') - key) % 26 + ord('a'))

            result += new_char

        else:
            result += char

    return result


# Test
message = "Hello World!"
key = 3

encoded = encode(message, key)
decoded = decode(encoded, key)

print("Original :", message)
print("Encoded  :", encoded)
print("Decoded  :", decoded)
```

### Task 12: Text Data Cleaning Pipeline

Build a `clean_dataset(records)` function that processes raw text records. Each record is a string with fields separated by `|`.

The function should:
1. Skip records where **any** field is empty
2. Clean names to **Title Case** and remove extra spaces
3. Validate that age is a number (18–120)
4. Validate email format (contains `@` and has a `.` after the `@`)
5. Standardise country names to **uppercase**
6. Return two lists: `valid_records` and `invalid_records` with reasons

**Methods to use:** `split()`, `strip()`, `title()`, `upper()`, `isdigit()`, `all()`, `find()`


```python
raw_data = [
    "  john SMITH  | 25 | john.smith@email.com | united kingdom ",
    "  MARY  johnson | abc | mary@email | france",
    " | 30 | test@test.com | germany",
    "  bob wilson  | 17 | bob@mail.com | spain  ",
    "  alice Brown  | 28 | alice.b@data.org | italy  ",
    "charlie DAVIS | 35 | | usa"
]

# Function to clean dataset
def clean_dataset(records):

    valid_records = []
    invalid_records = []

    for record in records:

        reasons = []

        # Split fields
        fields = record.split("|")

        # Remove spaces
        fields = [field.strip() for field in fields]

        # Check empty fields
        if not all(fields):

            reasons.append("Contains empty fields")

            invalid_records.append((record, reasons))

            continue

        # Store values
        name = fields[0].title()
        age = fields[1]
        email = fields[2]
        country = fields[3].upper()

        # Validate age
        if not age.isdigit():

            reasons.append("Age is not a number")

        else:

            age_number = int(age)

            if age_number < 18 or age_number > 120:

                reasons.append("Age must be between 18 and 120")

        # Validate email
        at_position = email.find("@")
        dot_position = email.find(".")

        if at_position == -1 or dot_position == -1 or dot_position < at_position:

            reasons.append("Invalid email format")

        # Valid record
        if reasons == []:

            valid_records.append({
                "Name": name,
                "Age": age,
                "Email": email,
                "Country": country
            })

        # Invalid record
        else:

            invalid_records.append((record, reasons))

    return valid_records, invalid_records


# Function call
valid, invalid = clean_dataset(raw_data)

print("VALID RECORDS:\n")

for item in valid:
    print(item)

print("\nINVALID RECORDS:\n")

for record, reason in invalid:
    print(record)
    print("Reasons:", reason)
    print()
```

### Task 13: Analyse & Report on Text Corpus

Write a function `text_report(text)` that produces a complete analysis:

1. **Basic stats:** Total characters, words, sentences, average word length
2. **Character breakdown:** Count of uppercase, lowercase, digits, spaces, punctuation
3. **Top 5 words:** Most frequently used words (exclude common words: 'the', 'a', 'is', 'and', 'of', 'to', 'in')
4. **Longest and shortest words**
5. **Print a formatted report** using string formatting

**Methods to use:** `split()`, `lower()`, `replace()`, `count()`, `isupper()`, `islower()`, `isdigit()`, `isspace()`, `join()`, `center()`, `ljust()`


```python
corpus = """Artificial Intelligence is transforming every industry. From healthcare to finance,
AI systems are making decisions that affect millions of people. Machine Learning,
a subset of AI, enables computers to learn from data without being explicitly programmed.
Deep Learning takes this further with neural networks inspired by the human brain.
In 2024, Large Language Models like GPT have shown remarkable abilities in understanding
and generating human language. The future of AI is both exciting and challenging."""

# Function for text analysis
def text_report(text):

    print(" TEXT ANALYSIS REPORT ".center(60, "="))

    # Basic stats
    total_characters = len(text)

    words = text.split()

    total_words = len(words)

    sentences = text.count(".") + text.count("!") + text.count("?")

    average_word_length = sum(len(word) for word in words) / total_words

    # Character breakdown
    uppercase = 0
    lowercase = 0
    digits = 0
    spaces = 0
    punctuation = 0

    punctuation_marks = ".,!?:;"

    for char in text:

        if char.isupper():
            uppercase += 1

        elif char.islower():
            lowercase += 1

        elif char.isdigit():
            digits += 1

        elif char.isspace():
            spaces += 1

        elif char in punctuation_marks:
            punctuation += 1

    # Clean words
    clean_text = text.lower()

    for mark in punctuation_marks:
        clean_text = clean_text.replace(mark, "")

    clean_words = clean_text.split()

    # Remove common words
    common_words = ['the', 'a', 'is', 'and', 'of', 'to', 'in']

    filtered_words = []

    for word in clean_words:

        if word not in common_words:
            filtered_words.append(word)

    # Word frequencies
    frequency = {}

    for word in filtered_words:
        frequency[word] = filtered_words.count(word)

    # Top 5 words
    sorted_words = sorted(frequency.items(),
                          key=lambda item: item[1],
                          reverse=True)

    # Longest and shortest words
    longest_word = max(filtered_words, key=len)

    shortest_word = min(filtered_words, key=len)

    # Print report
    print("\nBASIC STATS".ljust(30, "-"))

    print(f"Total Characters : {total_characters}")
    print(f"Total Words      : {total_words}")
    print(f"Total Sentences  : {sentences}")
    print(f"Average Word Len : {average_word_length:.2f}")

    print("\nCHARACTER BREAKDOWN".ljust(30, "-"))

    print(f"Uppercase Letters : {uppercase}")
    print(f"Lowercase Letters : {lowercase}")
    print(f"Digits            : {digits}")
    print(f"Spaces            : {spaces}")
    print(f"Punctuation       : {punctuation}")

    print("\nTOP 5 WORDS".ljust(30, "-"))

    for word, count in sorted_words[:5]:
        print(f"{word.ljust(15)} : {count}")

    print("\nWORD LENGTH ANALYSIS".ljust(30, "-"))

    print(f"Longest Word  : {longest_word}")
    print(f"Shortest Word : {shortest_word}")

    print("=" * 60)


# Function call
text_report(corpus)
```

### Task 14: Build a Simple Search Engine

Create a function `search(documents, query)` that:
1. Takes a list of document strings and a search query
2. Returns documents **ranked** by how many times the query words appear
3. The search should be **case-insensitive**
4. If query has multiple words, count matches for **each** word
5. Show a relevance score and highlight which query words were found

**Methods to use:** `lower()`, `split()`, `count()`, `any()`, `all()`, `strip()`, `replace()`


```python
documents = [
    "Python is a great programming language for data science and machine learning",
    "Data analysis with Python involves cleaning and processing large datasets",
    "Machine learning models can predict outcomes from historical data",
    "Web development with JavaScript and HTML is different from data work",
    "Python data science libraries include NumPy and Pandas for data processing"
]

# Search function
def search(documents, query):

    results = []

    # Clean query
    query = query.lower().strip()

    query_words = query.split()

    # Check each document
    for doc in documents:

        clean_doc = doc.lower()

        score = 0

        found_words = []

        # Count matches for each query word
        for word in query_words:

            count = clean_doc.count(word)

            if count > 0:

                score += count

                found_words.append(word)

        # Add matching documents only
        if score > 0:

            results.append((score, found_words, doc))

    # Sort by score (highest first)
    results.sort(reverse=True)

    # Print results
    print("\nSEARCH RESULTS\n")

    for score, found_words, doc in results:

        print(f"Score: {score}")

        print(f"Found Words: {', '.join(found_words)}")

        print(f"Document: {doc}")

        print("-" * 60)


# User query
query = "python data machine"

# Function call
search(documents, query)
```

### Task 9: Word Frequency Counter

Write a function `word_frequency(text)` that:
1. Converts text to lowercase
2. Removes punctuation (`.`, `,`, `!`, `?`)
3. Splits into words
4. Returns a **dictionary** of word → count
5. Print the top 5 most common words

**Methods to use:** `lower()`, `replace()`, `split()`, `count()`


```python
text = """Data science is the study of data. Data scientists use data to find patterns.
Machine learning is a part of data science. Scientists study data every day!
The science of data is growing, and data is everywhere."""

# Function to count word frequency
def word_frequency(text):

    # Convert text to lowercase
    text = text.lower()

    # Remove punctuation
    text = text.replace(".", "")
    text = text.replace(",", "")
    text = text.replace("!", "")
    text = text.replace("?", "")

    # Split text into words
    words = text.split()

    # Dictionary to store word counts
    frequency = {}

    # Count frequency of each word
    for word in words:

        frequency[word] = words.count(word)

    return frequency


# Function call
result = word_frequency(text)

print("Word Frequencies:\n")

# Print all word frequencies
for word, count in result.items():

    print(word, ":", count)


# Print top 5 most common words
print("\nTop 5 Most Common Words:\n")

sorted_words = sorted(result.items(),
                      key=lambda item: item[1],
                      reverse=True)

for word, count in sorted_words[:5]:

    print(word, ":", count)
```

### Task 10: Password Strength Checker

Write a function `check_password(password)` that scores a password (0–5):

| Rule | Points |
|------|--------|
| At least 8 characters | +1 |
| Contains uppercase letter | +1 |
| Contains lowercase letter | +1 |
| Contains a digit | +1 |
| Contains a special character (`@#$%!`) | +1 |

Print the score and a label: **Weak** (0-2), **Medium** (3-4), **Strong** (5).

If the password contains a space, print  and reject it.

**Methods to use:** `len()`, `isupper()`, `islower()`, `isdigit()`, `any()`


```python
test_passwords = [
    "hello",
    "Hello123",
    "My Pass!",
    "Str0ng@Pass",
    "12345678"
]

# Function to check password strength
def check_password(password):

    score = 0

    # Reject password with spaces
    if " " in password:

        print(password, "-> Invalid Password (contains spaces)")

        return

    # Check length
    if len(password) >= 8:
        score += 1

    # Check uppercase letter
    if any(char.isupper() for char in password):
        score += 1

    # Check lowercase letter
    if any(char.islower() for char in password):
        score += 1

    # Check digit
    if any(char.isdigit() for char in password):
        score += 1

    # Check special character
    if any(char in "@#$%!" for char in password):
        score += 1

    # Password strength
    if score <= 2:
        strength = "Weak"

    elif score <= 4:
        strength = "Medium"

    else:
        strength = "Strong"

    # Print result
    print(password, "-> Score:", score, "|", strength)


# Test all passwords
for pwd in test_passwords:

    check_password(pwd)
```

### Task 11: Caesar Cipher Encoder & Decoder

Write **two** functions:
1. `encode(text, key)` — shifts each letter forward by `key` positions
2. `decode(text, key)` — reverses the shift to get the original text

Rules:
- Only shift letters (a-z). Leave spaces, digits, and punctuation unchanged
- Wrap around: `z` shifted by 1 → `a`
- Preserve case: uppercase stays uppercase
- Verify: `decode(encode(text, key), key)` should return the original text

**Methods to use:** `ord()`, `chr()`, `isalpha()`, `isupper()`, `islower()`


```python
# Function to encode text
def encode(text, key):

    result = ""

    for char in text:

        # Check only letters
        if char.isalpha():

            # Uppercase letters
            if char.isupper():

                new_char = chr((ord(char) - ord('A') + key) % 26 + ord('A'))

            # Lowercase letters
            else:

                new_char = chr((ord(char) - ord('a') + key) % 26 + ord('a'))

            result += new_char

        # Keep spaces and symbols unchanged
        else:

            result += char

    return result


# Function to decode text
def decode(text, key):

    result = ""

    for char in text:

        if char.isalpha():

            if char.isupper():

                new_char = chr((ord(char) - ord('A') - key) % 26 + ord('A'))

            else:

                new_char = chr((ord(char) - ord('a') - key) % 26 + ord('a'))

            result += new_char

        else:

            result += char

    return result


# Test
message = "Hello World!"
key = 3

encoded = encode(message, key)

decoded = decode(encoded, key)

print("Original :", message)

print("Encoded  :", encoded)

print("Decoded  :", decoded)
```

### Task 12: Text Data Cleaning Pipeline

Build a `clean_dataset(records)` function that processes raw text records. Each record is a string with fields separated by `|`.

The function should:
1. Skip records where **any** field is empty
2. Clean names to **Title Case** and remove extra spaces
3. Validate that age is a number (18–120)
4. Validate email format (contains `@` and has a `.` after the `@`)
5. Standardise country names to **uppercase**
6. Return two lists: `valid_records` and `invalid_records` with reasons

**Methods to use:** `split()`, `strip()`, `title()`, `upper()`, `isdigit()`, `all()`, `find()`


```python
raw_data = [
    "  john SMITH  | 25 | john.smith@email.com | united kingdom ",
    "  MARY  johnson | abc | mary@email | france",
    " | 30 | test@test.com | germany",
    "  bob wilson  | 17 | bob@mail.com | spain  ",
    "  alice Brown  | 28 | alice.b@data.org | italy  ",
    "charlie DAVIS | 35 | | usa"
]

# Function to clean dataset
def clean_dataset(records):

    valid_records = []

    invalid_records = []

    # Check each record
    for record in records:

        reasons = []

        # Split record into fields
        fields = record.split("|")

        # Remove extra spaces
        fields = [field.strip() for field in fields]

        # Check empty fields
        if not all(fields):

            reasons.append("Contains empty fields")

            invalid_records.append((record, reasons))

            continue

        # Store fields
        name = fields[0].title()

        age = fields[1]

        email = fields[2]

        country = fields[3].upper()

        # Validate age
        if not age.isdigit():

            reasons.append("Age is not a number")

        else:

            age_number = int(age)

            if age_number < 18 or age_number > 120:

                reasons.append("Age must be between 18 and 120")

        # Validate email
        at_position = email.find("@")

        dot_position = email.find(".")

        if at_position == -1 or dot_position == -1 or dot_position < at_position:

            reasons.append("Invalid email format")

        # Store valid record
        if reasons == []:

            valid_records.append({
                "Name": name,
                "Age": age,
                "Email": email,
                "Country": country
            })

        # Store invalid record
        else:

            invalid_records.append((record, reasons))

    return valid_records, invalid_records


# Function call
valid, invalid = clean_dataset(raw_data)


# Print valid records
print("VALID RECORDS:\n")

for item in valid:

    print(item)


# Print invalid records
print("\nINVALID RECORDS:\n")

for record, reason in invalid:

    print(record)

    print("Reasons:", reason)

    print()
```

### Task 13: Analyse & Report on Text Corpus

Write a function `text_report(text)` that produces a complete analysis:

1. **Basic stats:** Total characters, words, sentences, average word length
2. **Character breakdown:** Count of uppercase, lowercase, digits, spaces, punctuation
3. **Top 5 words:** Most frequently used words (exclude common words: 'the', 'a', 'is', 'and', 'of', 'to', 'in')
4. **Longest and shortest words**
5. **Print a formatted report** using string formatting

**Methods to use:** `split()`, `lower()`, `replace()`, `count()`, `isupper()`, `islower()`, `isdigit()`, `isspace()`, `join()`, `center()`, `ljust()`


```python
corpus = """Artificial Intelligence is transforming every industry. From healthcare to finance,
AI systems are making decisions that affect millions of people. Machine Learning,
a subset of AI, enables computers to learn from data without being explicitly programmed.
Deep Learning takes this further with neural networks inspired by the human brain.
In 2024, Large Language Models like GPT have shown remarkable abilities in understanding
and generating human language. The future of AI is both exciting and challenging."""

# Function for text analysis
def text_report(text):

    print(" TEXT ANALYSIS REPORT ".center(60, "="))

    # Basic statistics
    total_characters = len(text)

    words = text.split()

    total_words = len(words)

    total_sentences = text.count(".") + text.count("!") + text.count("?")

    average_word_length = sum(len(word) for word in words) / total_words

    # Character breakdown
    uppercase = 0
    lowercase = 0
    digits = 0
    spaces = 0
    punctuation = 0

    punctuation_marks = ".,!?:;"

    for char in text:

        if char.isupper():
            uppercase += 1

        elif char.islower():
            lowercase += 1

        elif char.isdigit():
            digits += 1

        elif char.isspace():
            spaces += 1

        elif char in punctuation_marks:
            punctuation += 1

    # Clean text
    clean_text = text.lower()

    for mark in punctuation_marks:

        clean_text = clean_text.replace(mark, "")

    # Split words
    clean_words = clean_text.split()

    # Remove common words
    common_words = ['the', 'a', 'is', 'and', 'of', 'to', 'in']

    filtered_words = []

    for word in clean_words:

        if word not in common_words:

            filtered_words.append(word)

    # Word frequency
    frequency = {}

    for word in filtered_words:

        frequency[word] = filtered_words.count(word)

    # Top 5 words
    sorted_words = sorted(frequency.items(),
                          key=lambda item: item[1],
                          reverse=True)

    # Longest and shortest words
    longest_word = max(filtered_words, key=len)

    shortest_word = min(filtered_words, key=len)

    # Print report
    print("\nBASIC STATS".ljust(30, "-"))

    print(f"Total Characters : {total_characters}")

    print(f"Total Words      : {total_words}")

    print(f"Total Sentences  : {total_sentences}")

    print(f"Average Word Len : {average_word_length:.2f}")

    print("\nCHARACTER BREAKDOWN".ljust(30, "-"))

    print(f"Uppercase Letters : {uppercase}")

    print(f"Lowercase Letters : {lowercase}")

    print(f"Digits            : {digits}")

    print(f"Spaces            : {spaces}")

    print(f"Punctuation       : {punctuation}")

    print("\nTOP 5 WORDS".ljust(30, "-"))

    for word, count in sorted_words[:5]:

        print(f"{word.ljust(15)} : {count}")

    print("\nWORD LENGTH ANALYSIS".ljust(30, "-"))

    print(f"Longest Word  : {longest_word}")

    print(f"Shortest Word : {shortest_word}")

    print("=" * 60)


# Function call
text_report(corpus)
```

### Task 14: Build a Simple Search Engine

Create a function `search(documents, query)` that:
1. Takes a list of document strings and a search query
2. Returns documents **ranked** by how many times the query words appear
3. The search should be **case-insensitive**
4. If query has multiple words, count matches for **each** word
5. Show a relevance score and highlight which query words were found

**Methods to use:** `lower()`, `split()`, `count()`, `any()`, `all()`, `strip()`, `replace()`


```python
documents = [
    "Python is a great programming language for data science and machine learning",
    "Data analysis with Python involves cleaning and processing large datasets",
    "Machine learning models can predict outcomes from historical data",
    "Web development with JavaScript and HTML is different from data work",
    "Python data science libraries include NumPy and Pandas for data processing"
]

# Function to search documents
def search(documents, query):

    results = []

    # Clean query
    query = query.lower().strip()

    query_words = query.split()

    # Process each document
    for doc in documents:

        # Convert document to lowercase
        clean_doc = doc.lower()

        # Remove punctuation
        clean_doc = clean_doc.replace(".", "")
        clean_doc = clean_doc.replace(",", "")

        score = 0

        found_words = []

        # Count query words
        for word in query_words:

            count = clean_doc.count(word)

            if count > 0:

                score += count

                found_words.append(word)

        # Store matching documents
        if score > 0:

            results.append((score, found_words, doc))

    # Sort by highest score
    results.sort(reverse=True)

    # Print results
    print("\nSEARCH RESULTS\n")

    for score, found_words, doc in results:

        print(f"Relevance Score: {score}")

        print(f"Found Words: {', '.join(found_words)}")

        print(f"Document: {doc}")

        print("-" * 60)


# Search query
query = "python data machine"

# Function call
search(documents, query)
```
