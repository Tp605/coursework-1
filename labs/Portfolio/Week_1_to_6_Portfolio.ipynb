```python
import statistics as stats  # importing statistics module - i have "borrowing" built-in tools to calculate things.

# taken input from user about five numbers with spaces between them
user_input = input("Type five numbers with spaces between them: ")

# convert the input into a list of float numbers
data = list(map(float, user_input.split()))

# find total, average, middle value, and standard deviation
sum_value = sum(data)
average_value = stats.mean(data)
middle_value = stats.median(data)
spread_value = stats.stdev(data)

# show the final answers
print("Numbers entered:", " | ".join(map(str, data)))
print(f"Sum: {sum_value:.2f}")
print(f"Average: {average_value:.2f}")
print(f"Median: {middle_value:.2f}")
print(f"Standard Deviation: {spread_value:.2f}")
```

```python
def check_palindrome(text):
    # These line is used to convert the text into lowercase.
    text = text.lower()

    # I have done the reverse of the text using slicing
    opposite_text = text[::-1]

   # check if original text and reversed text are same
    if text == opposite_text:
        return True
    else:
        return False


# Taking input from the user to check if the word is a palindrome or not
user_word = input("Type a word: ")

# Call the function
answer = check_palindrome(user_word)

# Print the result
print(f"Is '{user_word}' a palindrome? {answer}")
```

***What things i used in above code are here:****


def and return — the function calculates and returns a result without printing anything, keeping the logic separate from the output

word.lower() — converts to lowercase first so "Racecar" and "racecar" are treated the same

Slicing [::-1] — a step of -1 with no start or stop index walks through the string backwards, producing the reversed version in one line

if / else returning True / False — the function gives back a boolean that the caller can use however they like

```python
# 2D list: [Student ID, Student Name, Math, Science, English]
class_records = [
    [101, "Alice", 85, 90, 88],
    [102, "Bob", 70, 65, 72],
    [103, "Charlie", 95, 92, 96],
    [104, "David", 78, 80, 76],
    [105, "Eva", 60, 55, 58],
]

# Function to calculate average marks
def find_average(record):
    total = record[2] + record[3] + record[4]
    avg_marks = total / 3
    return avg_marks

# Add average marks to each student record
for record in class_records:
    student_average = find_average(record)
    record.append(student_average)

# Display all student data
print("Updated Student Records:")
for record in class_records:
    print(record)

# Get all average marks using list comprehension
student_averages = [record[5] for record in class_records]

# Find highest and lowest average
maximum_average = max(student_averages)
minimum_average = min(student_averages)

print(f"\nMaximum Average: {maximum_average:.2f}")
print(f"Minimum Average: {minimum_average:.2f}")

# Find students who scored more than 75 in all subjects
good_students = [
    record[1]
    for record in class_records
    if record[2] > 75 and record[3] > 75 and record[4] > 75
]

print("\nStudents with more than 75 in every subject:")
print(good_students)
```

***What things i used in above code are here:****


2D list — each student is an inner list; you access individual values with student[2] for Math, student[3] for Science, etc.

Function with return — calculate_average does one job (calculate) and returns the result; it doesn't print anything, making it reusable

.append() on a nested list — adds the calculated average directly into each student's own inner list, extending the data in place

List comprehension — [student[5] for student in students] extracts just the averages into a flat list in one line

List comprehension with if — filters students by checking all three subject scores at once using and, giving only those who passed every subject

```python
# These names are both raw names with spaces and mixed uppercase/lowercase letters
employee_names = [
    "  TIRTH patel ",
    "mary JOHNSON",
    "  rishi KUMAR  ",
]

# Loop through each employee name
for employee in employee_names:
    # Remove extra spaces and make name proper format
    final_name = employee.strip().title()

    # Separate first name and last name
    split_name = final_name.split()

    # Create company email address
    first = split_name[0].lower()
    last = split_name[1].lower()
    employee_email = first + "." + last + "@company.com"

    # Display the final name and email
    print(final_name, "=", employee_email)
```

***CODE 2***

```python
# These are the log messages from a server, each with a timestamp and a log level (ERROR, INFO, WARNING).
server_logs = [
    "[14:32:05] ERROR: Disk full",
    "[14:32:10] INFO: Server started",
    "[14:33:01] WARNING: High memory usage",
    "[14:33:15] INFO: Backup complete",
    "[14:34:00] ERROR: Connection refused",
    "[14:35:22] WARNING: CPU at 90%"
]

def check_log_types(server_logs):
    # Counters is used to keep track of the number of each type of log message.
    total_errors = 0
    total_infos = 0
    total_warnings = 0

    for message in server_logs:
        message = message.strip()

        # Get timestamp from log message
        time_part = message.split("]")[0] + "]"
        print("Log time:", time_part)

        # Count each type of log
        if "ERROR" in message:
            total_errors += 1
        elif "INFO" in message:
            total_infos += 1
        elif "WARNING" in message:
            total_warnings += 1

    return total_errors, total_infos, total_warnings


# Store returned values
error_logs, info_logs, warning_logs = check_log_types(server_logs)

print("\nTotal number of ERROR logs:", error_logs)
print("Total number of INFO logs:", info_logs)
print("Total number of WARNING logs:", warning_logs)
```

***What things i used in above code are here:****


This example shows how string methods can parse structured text data like server logs. .split("]") breaks each line at the closing bracket to extract just the timestamp. The in operator checks whether a keyword like "ERROR" appears anywhere in the line, which is more flexible than checking exact positions. The function returns multiple values as a tuple, and they are unpacked in a single line. This is a practical pattern used in real monitoring tools and data pipelines.

```python
# This list is empty and will be used to store temperature values entered by the user.
daily_temps = []

# Create and write data into the file
with open("temps.txt", "w") as file:
    for number in range(1, 6):
        temperature = float(input(f"Enter Day {number} temperature: "))
        daily_temps.append(temperature)
        file.write(f"Day {number}: {temperature}°C\n")

# Read and display the saved file content
print("\n--- Saved Temperatures ---")
with open("temps.txt", "r") as file:
    for data in file:
        print(data, end="")

# This is used to calculate the average temperature from the list of daily temperatures.
total_temp = sum(daily_temps)
avg_temp = total_temp / len(daily_temps)

print(f"\nAverage temperature is: {avg_temp:.1f}°C")
```

***What things i used in above code are here:****



This covers the core of file handling — writing to a file with "w" mode, reading it back with "r" mode, and using the with statement to ensure the file closes automatically. It also shows how a list can hold data in memory while the file stores it permanently on disk. The f.write() call uses an f-string with \n to put each entry on its own line, and end="" in print() avoids extra blank lines when printing lines that already end with \n.

***CODE 2***

```python
# Dictionary where each category has a list of costs
spending = {
    "Food": [25, 40, 15],
    "Entertainment": [50, 20],
    "Travel": [300, 150]
}

def insert_cost(type_name, cost):
    # If category is not available, create empty list
    if type_name not in spending:
        spending[type_name] = []

    # Add cost into the category list
    spending[type_name].append(cost)
    print(f"${cost} added in {type_name}.")

def delete_cost(type_name, cost):
    # Check category and amount before removing
    if type_name in spending and cost in spending[type_name]:
        spending[type_name].remove(cost)
        print(f"${cost} removed from {type_name}.")
    else:
        print(f"${cost} was not found in {type_name}.")

def show_report():
    # Display total spending for each category
    print("\n--- Expense Report ---")

    final_total = 0

    for type_name, costs in spending.items():
        category_sum = sum(costs)
        final_total += category_sum
        print(f"{type_name}: ${category_sum}")

    print(f"Overall Total: ${final_total}")


# Testing the functions
insert_cost("Food", 30)
delete_cost("Travel", 150)
show_report()
```

***What things i used in above code are here:****



This covers the most important data structure of the week — a dictionary of lists. Each key is a category and each value is a list of amounts, which is a pattern used constantly in real data applications. It shows how to add a new key dynamically, how in checks membership, how .items() lets you loop over both keys and values at once, and how sum() works on a list stored inside a dictionary. Together with Example 1, these two codes cover both major topics of Week 5: file handling and data structures.
