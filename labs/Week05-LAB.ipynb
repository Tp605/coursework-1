### Activity 15: Managing Study Groups

You are responsible for managing a study group for a course. Each student can be part of multiple study groups, and you need to track which groups a student is enrolled in. Students can join or leave study groups. What data structure would you choose to manage this efficiently, and why?

**Example data:**

```python
{"John": ["Math", "Physics"], "Emily": ["Physics", "Biology"], "Alex": ["Math"]}

Steps:

1. Choose a data structure that allows quick access to the groups each student belongs to.
2. Write a program that lets students join and leave study groups.
3. Implement a function to display all groups a student is part of.
---



`````python
# Dictionary of lists: key = student name, value = list of groups
# Best choice because it allows O(1) lookup and supports multiple groups per student
study_groups = {
    "John": ["Math", "Physics"],
    "Emily": ["Physics", "Biology"],
    "Alex": ["Math"]
}

def join_group(student, group):
    # Add student if not already in the dictionary
    if student not in study_groups:
        study_groups[student] = []
    if group not in study_groups[student]:
        study_groups[student].append(group)
        print(f"{student} joined {group}.")
    else:
        print(f"{student} is already in {group}.")

def leave_group(student, group):
    # Remove group from student's list if it exists
    if student in study_groups and group in study_groups[student]:
        study_groups[student].remove(group)
        print(f"{student} left {group}.")
    else:
        print(f"{student} is not in {group}.")

def show_groups(student):
    # Display all groups the student belongs to
    if student in study_groups and study_groups[student]:
        print(f"{student}'s groups: {', '.join(study_groups[student])}")
    else:
        print(f"{student} is not in any groups.")

# Demo
join_group("Alex", "Biology")
leave_group("John", "Physics")
show_groups("John")
show_groups("Alex")
`````

`````python
import random

# A list is ideal: supports append, remove, and random.shuffle in-place
playlist = ["Song A", "Song B", "Song C", "Song D"]

def add_song(song):
    # Append the new song to the end of the playlist
    playlist.append(song)
    print(f"Added: {song}")

def remove_song(song):
    # Remove the song if it exists in the playlist
    if song in playlist:
        playlist.remove(song)
        print(f"Removed: {song}")
    else:
        print(f"{song} not found in playlist.")

def shuffle_playlist():
    # Shuffle the playlist randomly in place
    random.shuffle(playlist)
    print("Shuffled playlist:", playlist)

# Demo
print("Original:", playlist)
add_song("Song E")
remove_song("Song B")
shuffle_playlist()
`````

### Activity 20: Managing Event Tickets

You are developing a system to manage event tickets for concerts or sports events. Each event has a limited number of tickets available, and users can purchase tickets if they are still available. You need to track which users have purchased tickets for specific events and how many are still available. What data structure would you use for this task, and why?

**Example data:**
- Available tickets: `{"Concert": 100, "Football Game": 50, "Theater Show": 75}`
- Purchasers: `{"Concert": ["John", "Emily"], "Football Game": ["Alex"], "Theater Show": ["Sam", "John"]}`

**Steps:**
1. Choose a data structure to track ticket availability and purchasers efficiently.
2. Write a program that handles ticket purchases and updates the list of purchasers.
3. Implement functions to check availability and view who has purchased tickets.
---



`````python
# Two dictionaries:
# available_tickets: tracks remaining tickets per event
# purchasers: tracks who has bought tickets for each event
available_tickets = {"Concert": 100, "Football Game": 50, "Theater Show": 75}
purchasers        = {"Concert": ["John", "Emily"], "Football Game": ["Alex"], "Theater Show": ["Sam", "John"]}

def buy_ticket(user, event):
    if event not in available_tickets:
        print(f"Event '{event}' does not exist.")
        return
    if available_tickets[event] > 0:
        # Deduct one ticket and record the purchaser
        available_tickets[event] -= 1
        purchasers[event].append(user)
        print(f"{user} purchased a ticket for {event}. Remaining: {available_tickets[event]}")
    else:
        print(f"Sorry, no tickets left for {event}.")

def check_availability(event):
    # Show how many tickets are still available
    if event in available_tickets:
        print(f"{event}: {available_tickets[event]} ticket(s) available.")
    else:
        print(f"Event '{event}' not found.")

def view_purchasers(event):
    # Show who has bought tickets for the event
    if event in purchasers:
        print(f"Purchasers for {event}: {purchasers[event]}")
    else:
        print(f"Event '{event}' not found.")

# Demo
check_availability("Concert")
buy_ticket("Alice", "Concert")
view_purchasers("Concert")
`````

### Activity 21: Personal Finance App

You are creating a personal finance app that allows users to categorize their expenses and track spending by category (e.g., food, entertainment, travel). Users can add expenses, remove expenses, and see a summary of their spending. What data structure would you choose to store and organize the expenses, and why?

**Example data:**

```python
{"Food": [25, 40, 15], "Entertainment": [50, 20], "Travel": [300, 150]}
```
Steps:

1. Choose a data structure that lets users easily add and track expenses by category.
2. Write a program that allows users to input expenses, remove them, and view their total spending by category.
3. Implement a function to display a summary of all spending.
---


`````python
# Dictionary of lists: key = category, value = list of expense amounts
# Best choice — fast lookup by category, easy to append/remove/sum amounts
expenses = {"Food": [25, 40, 15], "Entertainment": [50, 20], "Travel": [300, 150]}

def add_expense(category, amount):
    # Create category if it doesn't exist, then add the amount
    if category not in expenses:
        expenses[category] = []
    expenses[category].append(amount)
    print(f"Added ${amount} to {category}.")

def remove_expense(category, amount):
    # Remove the first occurrence of the amount from the category
    if category in expenses and amount in expenses[category]:
        expenses[category].remove(amount)
        print(f"Removed ${amount} from {category}.")
    else:
        print(f"${amount} not found in {category}.")

def summary():
    # Print total spending per category and overall total
    print("\n--- Spending Summary ---")
    total = 0
    for category, amounts in expenses.items():
        cat_total = sum(amounts)
        total += cat_total
        print(f"  {category}: ${cat_total}")
    print(f"  TOTAL: ${total}")

# Demo
add_expense("Food", 30)
remove_expense("Travel", 150)
summary()
`````

Write a function `parse_csv_to_dicts(filepath)` that reads any comma-separated file and returns a list of dictionaries.
Then use your function to read `week05_data/sample_data.csv` and perform some data analysis!


`````python
def parse_csv_to_dicts(filepath):
    """Read a CSV file and return a list of dicts keyed by header names."""
    result = []
    with open(r"C:\Users\HP\OneDrive\Desktop\1510 python\labs\sample_data.csv", "r") as f:
        lines = f.readlines()
    if not lines:
        return result
    
    # First line is the header row
    headers = [h.strip() for h in lines[0].split(",")]
    
    # Each remaining line becomes a dictionary
    for line in lines[1:]:
        line = line.strip()
        if not line:
            continue
        values = [v.strip() for v in line.split(",")]
        row = dict(zip(headers, values))
        result.append(row)
    return result


# Load the CSV file
data = parse_csv_to_dicts(r"C:\Users\HP\OneDrive\Desktop\1510 python\labs\sample_data.csv")

print(f"Loaded {len(data)} rows.")
print("Columns:", list(data[0].keys()) if data else "(no data)")

# Basic EDA: find numeric columns and compute min / max / average
if data:
    numeric_cols = {}
    for col in data[0]:
        try:
            values = [float(row[col]) for row in data if row[col]]
            numeric_cols[col] = values
        except ValueError:
            pass  # skip non-numeric columns

    print("\n--- Numeric Column Statistics ---")
    for col, values in numeric_cols.items():
        print(f"  {col}: min={min(values):.2f}, max={max(values):.2f}, avg={sum(values)/len(values):.2f}")
`````
